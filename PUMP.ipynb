```python
#!/usr/bin/env python3
"""
Stock Watchlist Dashboard
=========================
Last updated: 2026-06-08

To update: Ask Ona to "analyse the latest news and update stock_watchlist.py"
She will refresh prices, metrics, catalysts, and potential estimates.
"""

from datetime import date

# ============================================================================
# CONFIG — Update this date each time the table is refreshed
# ============================================================================
LAST_UPDATED = "2026-06-08"

# ============================================================================
# STOCK DATA — Each entry is a dict. Edit/add/remove entries as needed.
#
# Fields:
#   ticker, company, sector, price, mkt_cap, rev_ttm, rev_growth,
#   net_income, eps, pe, gross_margin, profit_margin, fcf_margin,
#   w52_low, w52_high, pumped, analyst_rating, price_target, profitable,
#   catalyst, risk, pot_1y, pot_2y, pot_5y
#
# pumped values : "YES", "PARTIAL", "NO"
# risk values   : "Low", "Medium", "High", "Very High"
# pot_Xy        : estimated potential % change (str like "+50%" or "-20% to +80%")
# ============================================================================

STOCKS = [
    {
        "ticker": "BE",
        "company": "Bloom Energy",
        "sector": "Energy",
        "price": 264,
        "mkt_cap": "75B",
        "rev_ttm": "2.45B",
        "rev_growth": "+56.5%",
        "net_income": "6M",
        "eps": 0.02,
        "pe": "11,067x",
        "gross_margin": "—",
        "profit_margin": "~0.2%",
        "fcf_margin": "—",
        "w52_low": 20,
        "w52_high": 323,
        "pumped": "YES (~13x)",
        "analyst_rating": "Buy",
        "price_target": 263,
        "profitable": "Barely",
        "catalyst": "Oracle/Brookfield data center deals",
        "risk": "High",
        "pot_1y": "-20% to +30%",
        "pot_2y": "-10% to +50%",
        "pot_5y": "+0% to +100%",
    },
    {
        "ticker": "CRDO",
        "company": "Credo Technology",
        "sector": "Semiconductors",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "1.34B",
        "rev_growth": "+206%",
        "net_income": "472M",
        "eps": 2.51,
        "pe": "—",
        "gross_margin": "68%",
        "profit_margin": "35.4%",
        "fcf_margin": "~0%",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "YES (~5x)",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "Yes",
        "catalyst": "80%+ FY27 revenue guidance, AI connectivity",
        "risk": "High",
        "pot_1y": "-10% to +40%",
        "pot_2y": "+10% to +80%",
        "pot_5y": "+50% to +200%",
    },
    {
        "ticker": "VRT",
        "company": "Vertiv Holdings",
        "sector": "Industrials",
        "price": 301,
        "mkt_cap": "115B",
        "rev_ttm": "10.8B",
        "rev_growth": "+29%",
        "net_income": "1.56B",
        "eps": 3.98,
        "pe": "75.5x",
        "gross_margin": "—",
        "profit_margin": "14.4%",
        "fcf_margin": "—",
        "w52_low": 107,
        "w52_high": 380,
        "pumped": "YES (~3x)",
        "analyst_rating": "Buy",
        "price_target": 377,
        "profitable": "Yes",
        "catalyst": "AI data center power & cooling",
        "risk": "Medium",
        "pot_1y": "-10% to +30%",
        "pot_2y": "+0% to +60%",
        "pot_5y": "+30% to +150%",
    },
    {
        "ticker": "FIX",
        "company": "Comfort Systems",
        "sector": "Industrials",
        "price": 1844,
        "mkt_cap": "64.8B",
        "rev_ttm": "10.1B",
        "rev_growth": "+38.4%",
        "net_income": "1.22B",
        "eps": 34.64,
        "pe": "53.2x",
        "gross_margin": "25.1%",
        "profit_margin": "12.1%",
        "fcf_margin": "~0%",
        "w52_low": 469,
        "w52_high": 2074,
        "pumped": "YES (~4x)",
        "analyst_rating": "Strong Buy",
        "price_target": 2026,
        "profitable": "Yes",
        "catalyst": "Data center HVAC & electrical buildout",
        "risk": "Medium",
        "pot_1y": "-10% to +20%",
        "pot_2y": "+0% to +40%",
        "pot_5y": "+20% to +100%",
    },
    {
        "ticker": "RKLB",
        "company": "Rocket Lab",
        "sector": "Aerospace",
        "price": 110,
        "mkt_cap": "63.7B",
        "rev_ttm": "680M",
        "rev_growth": "+45.8%",
        "net_income": "-183M",
        "eps": -0.33,
        "pe": "N/A",
        "gross_margin": "—",
        "profit_margin": "Neg",
        "fcf_margin": "—",
        "w52_low": 25,
        "w52_high": 151,
        "pumped": "YES (~4x)",
        "analyst_rating": "Buy",
        "price_target": 105,
        "profitable": "No",
        "catalyst": "Neutron rocket, satellite constellation",
        "risk": "High",
        "pot_1y": "-30% to +40%",
        "pot_2y": "-10% to +100%",
        "pot_5y": "+50% to +300%",
    },
    {
        "ticker": "POWL",
        "company": "Powell Industries",
        "sector": "Industrials",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "1.13B",
        "rev_growth": "+4.5%",
        "net_income": "187M",
        "eps": 5.11,
        "pe": "—",
        "gross_margin": "30.1%",
        "profit_margin": "16.5%",
        "fcf_margin": "17%",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "PARTIAL (~2-3x)",
        "analyst_rating": "Overweight",
        "price_target": 310,
        "profitable": "Yes",
        "catalyst": "$1.8B backlog, record $400M data center order",
        "risk": "Medium",
        "pot_1y": "+10% to +50%",
        "pot_2y": "+20% to +100%",
        "pot_5y": "+50% to +200%",
    },
    {
        "ticker": "LSCC",
        "company": "Lattice Semiconductor",
        "sector": "Semiconductors",
        "price": 136,
        "mkt_cap": "18.6B",
        "rev_ttm": "574M",
        "rev_growth": "+17.4%",
        "net_income": "20M",
        "eps": 0.14,
        "pe": "961x",
        "gross_margin": "68.4%",
        "profit_margin": "3.5%",
        "fcf_margin": "26%",
        "w52_low": 46,
        "w52_high": 156,
        "pumped": "PARTIAL (~3x from low)",
        "analyst_rating": "Strong Buy",
        "price_target": 147,
        "profitable": "Barely",
        "catalyst": "Cyclical recovery, edge AI, automotive",
        "risk": "Medium",
        "pot_1y": "+10% to +50%",
        "pot_2y": "+30% to +120%",
        "pot_5y": "+80% to +300%",
    },
    {
        "ticker": "LEU",
        "company": "Centrus Energy",
        "sector": "Energy",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "452M",
        "rev_growth": "~0%",
        "net_income": "61M",
        "eps": 3.02,
        "pe": "—",
        "gross_margin": "25.7%",
        "profit_margin": "13.4%",
        "fcf_margin": "-13.6%",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "NO",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "Yes",
        "catalyst": "$900M HALEU contract, monopoly nuclear fuel supplier",
        "risk": "Medium",
        "pot_1y": "+20% to +80%",
        "pot_2y": "+50% to +200%",
        "pot_5y": "+100% to +500%",
    },
    {
        "ticker": "NBIS",
        "company": "Nebius Group",
        "sector": "Technology",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "878M",
        "rev_growth": "+453%",
        "net_income": "918M*",
        "eps": 3.51,
        "pe": "—",
        "gross_margin": "83.9%",
        "profit_margin": "95%*",
        "fcf_margin": "-359%",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "NO",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "Mixed*",
        "catalyst": "Aschenbrenner stake, GPU cloud for AI",
        "risk": "High",
        "pot_1y": "-20% to +100%",
        "pot_2y": "+0% to +200%",
        "pot_5y": "+50% to +500%",
    },
    {
        "ticker": "FCEL",
        "company": "FuelCell Energy",
        "sector": "Energy",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "170M",
        "rev_growth": "+48.3%",
        "net_income": "-185M",
        "eps": -6.91,
        "pe": "N/A",
        "gross_margin": "-16%",
        "profit_margin": "Neg",
        "fcf_margin": "-77.6%",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "NO",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "No",
        "catalyst": "1.5 GW data center pipeline, carbon capture",
        "risk": "High",
        "pot_1y": "-40% to +100%",
        "pot_2y": "-20% to +200%",
        "pot_5y": "+0% to +500%",
    },
    {
        "ticker": "PLUG",
        "company": "Plug Power",
        "sector": "Energy",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "740M",
        "rev_growth": "+15.2%",
        "net_income": "-1.68B",
        "eps": -1.38,
        "pe": "N/A",
        "gross_margin": "-25.7%",
        "profit_margin": "Neg",
        "fcf_margin": "-88.3%",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "PARTIAL (+330% 12mo)",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "No",
        "catalyst": "EBITDA+ target Q4 2026, hydrogen infra",
        "risk": "Very High",
        "pot_1y": "-50% to +100%",
        "pot_2y": "-30% to +200%",
        "pot_5y": "-50% to +500%",
    },
    {
        "ticker": "OKLO",
        "company": "Oklo Inc",
        "sector": "Energy",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "0",
        "rev_growth": "Pre-rev",
        "net_income": "-129M",
        "eps": -0.84,
        "pe": "N/A",
        "gross_margin": "N/A",
        "profit_margin": "N/A",
        "fcf_margin": "N/A",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "NO",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "No",
        "catalyst": "DOE selection, Meta partnership, Sam Altman backed",
        "risk": "Very High",
        "pot_1y": "-40% to +100%",
        "pot_2y": "-20% to +300%",
        "pot_5y": "+0% to +1000%",
    },
    {
        "ticker": "SMR",
        "company": "NuScale Power",
        "sector": "Energy",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "19M",
        "rev_growth": "-62%",
        "net_income": "-681M",
        "eps": -2.92,
        "pe": "N/A",
        "gross_margin": "23.8%",
        "profit_margin": "Neg",
        "fcf_margin": "-4036%",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "NO",
        "analyst_rating": "Neutral",
        "price_target": 12,
        "profitable": "No",
        "catalyst": "Only NRC-approved SMR design, TVA project",
        "risk": "Very High",
        "pot_1y": "-50% to +80%",
        "pot_2y": "-30% to +200%",
        "pot_5y": "+0% to +800%",
    },
    {
        "ticker": "VKTX",
        "company": "Viking Therapeutics",
        "sector": "Healthcare",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "0",
        "rev_growth": "Pre-rev",
        "net_income": "—",
        "eps": "—",
        "pe": "N/A",
        "gross_margin": "N/A",
        "profit_margin": "N/A",
        "fcf_margin": "N/A",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "NO",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "No",
        "catalyst": "Phase III obesity drug enrolled, data 2027",
        "risk": "Very High",
        "pot_1y": "-50% to +100%",
        "pot_2y": "-30% to +300%",
        "pot_5y": "+0% to +1000%",
    },
    {
        "ticker": "ZNTL",
        "company": "Zentalis Pharma",
        "sector": "Healthcare",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "0",
        "rev_growth": "Pre-rev",
        "net_income": "-124M",
        "eps": -1.73,
        "pe": "N/A",
        "gross_margin": "N/A",
        "profit_margin": "N/A",
        "fcf_margin": "N/A",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "NO",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "No",
        "catalyst": "Azenosertib pivotal trials, ovarian cancer",
        "risk": "Very High",
        "pot_1y": "-60% to +100%",
        "pot_2y": "-40% to +300%",
        "pot_5y": "-80% to +500%",
    },
    {
        "ticker": "BEP",
        "company": "Brookfield Renewable",
        "sector": "Energy",
        "price": "—",
        "mkt_cap": "—",
        "rev_ttm": "6.34B",
        "rev_growth": "+6.3%",
        "net_income": "525M",
        "eps": -0.31,
        "pe": "N/A",
        "gross_margin": "48.7%",
        "profit_margin": "8.3%",
        "fcf_margin": "-85%",
        "w52_low": "—",
        "w52_high": "—",
        "pumped": "NO",
        "analyst_rating": "—",
        "price_target": "—",
        "profitable": "Mixed",
        "catalyst": "Hyperscaler PPAs, Boralex acquisition",
        "risk": "Low",
        "pot_1y": "+5% to +20%",
        "pot_2y": "+10% to +40%",
        "pot_5y": "+30% to +80%",
    },
]


# ============================================================================
# DISPLAY
# ============================================================================

def fmt(val, width):
    """Right-pad a string value to a fixed width."""
    s = str(val) if val is not None else "—"
    return s.ljust(width)


def print_table(stocks):
    """Print the watchlist as a formatted table."""

    # Column definitions: (header, dict key, width)
    cols = [
        ("Ticker",       "ticker",         7),
        ("Company",      "company",        22),
        ("Sector",       "sector",         15),
        ("Price",        "price",          8),
        ("Mkt Cap",      "mkt_cap",        8),
        ("Rev TTM",      "rev_ttm",        9),
        ("Rev Grw",      "rev_growth",     9),
        ("Net Inc",      "net_income",     9),
        ("EPS",          "eps",            7),
        ("PE",           "pe",             10),
        ("GM",           "gross_margin",   8),
        ("PM",           "profit_margin",  8),
        ("FCF M",        "fcf_margin",     8),
        ("52W Lo",       "w52_low",        7),
        ("52W Hi",       "w52_high",       7),
        ("Pumped?",      "pumped",         20),
        ("Rating",       "analyst_rating", 12),
        ("PT",           "price_target",   6),
        ("Profit?",      "profitable",     8),
        ("Risk",         "risk",           10),
        ("Pot 1Y",       "pot_1y",         16),
        ("Pot 2Y",       "pot_2y",         16),
        ("Pot 5Y",       "pot_5y",         18),
        ("Catalyst",     "catalyst",       50),
    ]

    # Header
    header = " | ".join(fmt(c[0], c[2]) for c in cols)
    sep = "-+-".join("-" * c[2] for c in cols)

    print()
    print(f"  STOCK WATCHLIST — Last updated: {LAST_UPDATED}")
    print(f"  {'=' * len(header)}")
    print(f"  {header}")
    print(f"  {sep}")

    # Rows
    for s in stocks:
        row = " | ".join(fmt(s.get(c[1], "—"), c[2]) for c in cols)
        print(f"  {row}")

    print(f"  {sep}")
    print()

    # ---- Footnotes ----
    print("  FOOTNOTES")
    print("  " + "=" * 80)
    print()
    print("  Column Abbreviations:")
    print("  ─────────────────────")
    print("  Ticker     = Stock ticker symbol")
    print("  Mkt Cap    = Market Capitalization (total value of outstanding shares)")
    print("  Rev TTM    = Revenue, Trailing Twelve Months")
    print("  Rev Grw    = Revenue Growth, Year-over-Year")
    print("  Net Inc    = Net Income, Trailing Twelve Months")
    print("  EPS        = Earnings Per Share (net income / shares outstanding)")
    print("  PE         = Price-to-Earnings ratio (price / EPS)")
    print("  GM         = Gross Margin (gross profit / revenue)")
    print("  PM         = Profit Margin (net income / revenue)")
    print("  FCF M      = Free Cash Flow Margin (free cash flow / revenue)")
    print("  52W Lo/Hi  = 52-Week Low / High price")
    print("  Pumped?    = Whether the stock already had a major multi-bagger run")
    print("  Rating     = Analyst consensus rating")
    print("  PT         = Analyst consensus Price Target")
    print("  Profit?    = Whether the company is currently profitable")
    print("  Risk       = Risk level: Low / Medium / High / Very High")
    print("  Pot 1Y     = Estimated potential price change over 1 year")
    print("  Pot 2Y     = Estimated potential price change over 2 years")
    print("  Pot 5Y     = Estimated potential price change over 5 years")
    print()
    print("  Notes:")
    print("  ──────")
    print("  * NBIS net income includes large one-time non-operating gains;")
    print("    core operations still burning cash heavily (-$3.2B FCF).")
    print("  * Potential estimates (Pot 1Y/2Y/5Y) are rough directional ranges,")
    print("    NOT predictions. They reflect bull/bear scenarios based on current")
    print("    catalysts and fundamentals. Actual results may fall outside these ranges.")
    print("  * 'Pre-rev' = pre-revenue company (no product sales yet).")
    print("  * N/A = not applicable (e.g., PE ratio for unprofitable companies).")
    print("  * '—' = data not available at time of last update.")
    print()
    print("  Pumped Status Legend:")
    print("  ─────────────────────")
    print("  YES        = Already had a major run (likely late to enter)")
    print("  PARTIAL    = Has moved significantly but may have more room")
    print("  NO         = Hasn't had its breakout move yet")
    print()
    print("  ⚠️  DISCLAIMER: This is NOT financial advice. Data is sourced from")
    print("  public sources and may be outdated. Always do your own due diligence.")
    print("  Past performance does not guarantee future results.")
    print()


if __name__ == "__main__":
    print_table(STOCKS)
```
